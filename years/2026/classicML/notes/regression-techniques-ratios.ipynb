## Complete Before/After Examples for All Parameters

### 1. Normalization (`normalize = True`)
**MinMaxScaler: scales to 0-1 range**

```
BEFORE Carat Weight: [0.2, 0.5, 1.8, 2.1]
AFTER Normalized:    [0.00, 0.15, 0.82, 1.00]
```
*Formula*: `(x - min) / (max - min)`

### 2. Transformation (`transformation = True`)
**Yeo-Johnson: fixes skewness**

```
BEFORE Price: [100, 200, 500, 8000]
Skewness: ~2.1 (highly right-skewed)

AFTER Transformed: [~4.62, ~5.30, ~6.21, ~9.00]
Skewness: ~0.3 (nearly normal)
```

### 3. Target Transform (`transform_target = True`)
**ONLY applies to Price column**

```
Dataset BEFORE:
Carat: [0.2, 0.5, 1.8]  ← unchanged
Price: [100, 500, 8000] ← gets Yeo-Johnson

Dataset AFTER:
Carat: [0.2, 0.5, 1.8]  ← still unchanged  
Price: [4.62, 6.21, 9.00] ← transformed
```

### 4. Multicollinearity (`remove_multicollinearity = True`)
**Removes features correlated > 0.95**

```
BEFORE (3 features):
Carat Weight: [0.2, 0.5, 1.8]
Weight Grams: [0.4, 1.0, 3.6]  ← corr=0.98 with Carat
Length: [4.1, 4.5, 6.2]       ← corr=0.87 (kept)

AFTER (2 features):
Carat Weight: [0.2, 0.5, 1.8]  ← kept
Length: [4.1, 4.5, 6.2]        ← kept
Weight Grams: DROPPED           ← too correlated
```

### 5. Binning (`bin_numeric_features = ['Carat Weight']`)
**Continuous → discrete categories**

```
BEFORE Carat: [0.23, 0.45, 1.2, 1.8, 2.1]
AFTER Bins:   ['0-0.5', '0-0.5', '1-1.5', '1.5-2', '2-2.5']
```

## Pipeline Sequence
```
Raw Data 
↓ normalize (scale numbers)
↓ transformation (fix skewness) 
↓ multicollinearity (drop correlated) 
↓ binning (discretize Carat)
↓ READY for modeling
```

Each step builds on the previous transformation, creating an optimized dataset for ML models.

## Regression Metrics Explained

These metrics evaluate model performance from `compare_models()` in PyCaret. **Lower is better** for MAE, MSE, RMSE, RMSLE, MAPE. **Higher is better** for R².

| Metric | Full Name | What It Measures | Example |
|--------|-----------|------------------|---------|
| **MAE** | Mean Absolute Error | Average prediction error (absolute difference) | Pred: 500, Actual: 600 → error = 100 |
| **MSE** | Mean Squared Error | Average squared error (penalizes large errors more) | Error 100 → 10,000 (squared) |
| **RMSE** | Root Mean Squared Error | MSE in original units (most interpretable) | √10,000 = 100 (same scale as Price) |
| **R²** | R-squared | % of variance explained (0-1, closer to 1 = better) | 0.85 = 85% of Price variation explained |
| **RMSLE** | Root Mean Squared Log Error | Error on log scale (good for skewed targets like Price) | log(500) vs log(600) → smaller relative errors |
| **MAPE** | Mean Absolute Percentage Error | Error as % of actual values | (100/600) × 100 = 16.7% error |
| **TT (Sec)** | Training Time | Seconds to train model (faster = better for comparison) | LightGBM: 0.5s vs RandomForest: 2.3s |

## Practical Interpretation
```
Best model might have:
MAE: 450          → Average $450 off on price predictions
RMSE: 650         → Typical error $650 
R²: 0.92         → Explains 92% of price variation
MAPE: 12%        → 12% average error relative to actual prices
RMSLE: 0.11      → Good for log-normal price distribution
```

1. **RMSE** - Most important (same units as Price)
2. **R²** - How much variation model captures  
3. **MAPE** - Business-relevant % error
4. **MAE** - Average dollar error

## Simple Examples for Each Metric

**Sample Data**: Predicting diamond prices
```
Actual Prices:  [500, 1200, 800, 2000]
Predicted:     [450, 1300, 850, 1900]
Errors:        [50, 100, 50, 100]
```

### 1. **MAE (Mean Absolute Error)**
**Average dollar mistake**
```
Errors: 50 + 100 + 50 + 100 = 300
MAE = 300 ÷ 4 = **75**
```
*Meaning*: Model wrong by **$75 on average**

### 2. **MSE (Mean Squared Error)** 
**Punishes big mistakes more**
```
Squared errors: 2500 + 10000 + 2500 + 10000 = 25000
MSE = 25000 ÷ 4 = **6250**
```
*Meaning*: Large errors hurt more (100² = 10,000)

### 3. **RMSE (Root Mean Squared Error)**
**Back to dollars, like MAE**
```
RMSE = √6250 = **79**
```
*Meaning*: **Typical error $79** (most useful for business)

### 4. **R² (R-squared)**
**How much price variation model explains**
```
Perfect predictions → R² = 1.0 (100%)
Random guessing    → R² = 0.0 (0%)
Your model: **0.92** = **92% explained**
```
*Meaning*: Model captures **92% of price patterns**

### 5. **RMSLE (Root Mean Squared Log Error)**
**Good for skewed prices (many cheap, few expensive diamonds)**
```
log(500)≈6.21, log(450)≈6.11 → small log error
log(2000)≈7.60, log(1900)≈7.55 → small log error
RMSLE = **0.08**
```
*Meaning*: **8% relative error** (ignores scale)

### 6. **MAPE (Mean Absolute Percentage Error)**
**Error as percentage**
```
% errors: (50/500)=10%, (100/1200)=8%, (50/800)=6%, (100/2000)=5%
MAPE = (10+8+6+5)÷4 = **7.25%**
```
*Meaning*: Wrong by **7% of actual price on average**

### 7. **TT (Sec) (Training Time)**
```
LightGBM: **0.5 seconds**
Random Forest: **2.3 seconds**
```
*Meaning*: How long to train (faster = better for testing)
