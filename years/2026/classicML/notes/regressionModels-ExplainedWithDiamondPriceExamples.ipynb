## Regression Models Explained with Diamond Price Examples
(regressionModels-ExplainedWithDiamondPriceExamples)
**Sample Data**: 5 diamonds with Carat, Cut, Color features → Predict Price

```
Carat  Cut  Color  Actual Price
0.5    Good D      1200
1.2    Fair E      2500  
2.0    Ideal D     8000
0.8    Good F      1800
1.5    Premium E   4500
```

***

### **Tree-Based Models** (Best for diamonds - handle non-linear patterns)

**1. Decision Tree Regressor**
```
Splits data like a flowchart:
Carat < 1.0? → Price ~1500
       No?   → Carat < 1.8? → Price ~4000  
                No? → Price ~7000
```
**Simple but overfits** (memorizes training data)

**2. Random Forest Regressor** ⭐ **TOP PERFORMER**
```
100 Decision Trees vote together
Tree1: $1250    Tree67: $1180
Tree2: $1220    Tree89: $1230
...
Average prediction: **$1205**
```
**Strong, stable, handles diamonds perfectly**

**3. Extra Trees Regressor**
```
Like Random Forest but random-er splits
Faster training, similar accuracy
```

**4. Gradient Boosting Regressor**
```
Trees learn from each other's mistakes
Tree1 predicts → Tree2 fixes errors → Tree3 fixes remaining
**Very accurate but slower**
```

**5. Light Gradient Boosting (LightGBM)** ⭐ **FASTEST + ACCURATE**
```
Super-optimized Gradient Boosting
Handles diamonds best in your table (lowest RMSE)
```

***

### **Linear Models** (Simple straight lines)

**6. Linear Regression**
```
Price = 2000 × Carat + 300 × Cut_score - 100 × Color_score
0.5 carat → 2000×0.5 = **$1000**
```
**Fast but assumes straight-line relationship**

**7. Ridge Regression**
```
Linear + shrinks big coefficients
Prevents overfitting on multicollinear features (Carat vs Length)
```

**8. Lasso Regression** 
```
Linear + sets useless features to exactly 0
Automatically feature selection
```

**9. Elastic Net**
```
Ridge + Lasso combined (best of both)
```

**10. Bayesian Ridge**
```
Linear with uncertainty estimates
Good when you don't trust data completely
```

***

### **Specialized Models**

**11. Huber Regressor**
```
Linear but gentle with outliers
Ignores crazy expensive diamonds
```

**12. AdaBoost Regressor**
```
Weak learners (tiny trees) boosted together
Slower than LightGBM
```

**13. K Neighbors Regressor (KNN)**
```
"Similar diamonds had these prices"
Finds 5 nearest neighbors → averages their prices
Slow on large datasets
```

**14. Orthogonal Matching Pursuit (OMP)**
```
Selects most important features step-by-step
Sparse models (few features)
```

**15. Lasso Least Angle Regression (LARS)**
```
Fast way to find Lasso solution
```

**16. Least Angle Regression (LAR)**
```
Similar to LARS but no shrinkage
```

**17. Dummy Regressor**
```
Baseline - always predicts average price ($3600)
Shows if your model beats "do nothing"
```

### **3. Extra Trees Regressor**
```
Like Random Forest but picks split points randomly
Tree1: Carat<1.3 → $2400
Tree2: Color=E → $2550  
100 trees average: **$2480**
```
**Faster than Random Forest, similar accuracy**

### **4. Decision Tree Regressor** 
```
Single tree decision process:
Carat > 1.0? → YES
Cut = Fair? → YES  
Color = E? → YES
→ Leaf predicts **$2500**
```
**Simple but overfits (learns training data too well)**

### **5. Gradient Boosting Regressor**
```
Tree1 predicts: $2000 (error +500)
Tree2 learns only the +500 error: +450  
Tree3 learns remaining +50 error: +40
Final: 2000 + 450 + 40 = **$2490**
```
**Very accurate, learns sequentially**

### **6. Light Gradient Boosting Machine (LightGBM)** ⭐
```
Optimized Gradient Boosting:
Histograms + leaf-wise growth = super fast
Predicts **$2495** in 0.5 seconds
```
**Your table winner - fastest + most accurate**

### **7. Huber Regressor**
```
Like linear but ignores outliers:
Price = 2100×Carat + adjustments
Ignores one $15,000 outlier diamond
→ **$2480**
```

### **8. Ridge Regression**
```
Price = 2050×Carat - 20×Color_score + ...
Shrinks Carat coefficient from 2200→2050
→ **$2470**
```

### **9. Bayesian Ridge**
```
Price = 2080×Carat ± uncertainty
"95% confident prediction is $2470-$2520"
```

### **10. Linear Regression**
```
Simple formula: Price = 2000×Carat + 200×Cut_score
1.2 carat Fair E → **$2460**
```
**Fastest but assumes straight line**

### **11. AdaBoost Regressor**
```
50 weak stumps boosted:
Stump1: Carat>1? → +$500
Stump2: Color=E? → +$200  
Final boost: **$2485**
```

### **12. K Neighbors Regressor (KNN)**
```
Finds 5 most similar diamonds:
Neighbor1: $2550, N2: $2450, N3: $2600...
Average 5 neighbors: **$2510**
```
**Slow on many diamonds**

### **13. Lasso Regression**
```
Price = 2100×Carat + 0×Cut + 0×Color (dropped useless features)
→ **$2520**
```

### **14. Elastic Net**
```
Price = 2070×Carat + 10×Cut - 5×Color
Combines Lasso + Ridge shrinkage
→ **$2490**
```

### **15. Lasso Least Angle Regression (LARS Lasso)**
```
Fast Lasso algorithm
Same as Lasso: **$2520**
```

### **16. Dummy Regressor**
```
Always predicts average price: **$3600**
```
**Baseline - beat this or model is useless**

### **17. Least Angle Regression (LAR)**

## LightGBM Hyperparameters Explained

```python
lgbm_params = {
    'num_leaves': np.arange(10,200,10),     # [10,20,30,...,190]
    'max_depth': [int(x) for x in np.linspace(10, 110, num = 11)],  # [10,20,30,...,110]
    'learning_rate': np.arange(0.1,1,0.1)   # [0.1,0.2,0.3,...,0.9]
}
```

## Each Parameter with Sample Data

**Sample Diamond**: Carat=1.2, Cut=Fair, Color=E → Actual Price=$2500

### 1. **`num_leaves`: [10, 20, 30, ..., 190]`**
**How many "decision points" per tree**

```
num_leaves=10  → Simple tree (10 leaves)
               Tree looks like:
               Carat<1.0? → $1500
               Carat>2.0? → $7000
               Else → **$2500**

num_leaves=100 → Complex tree (100 leaves)
                Can distinguish: 1.2 carat + Fair + E = **$2490**
```
**Small=simple, fast | Large=accurate, slow**

### 2. **`max_depth`: [10, 20, 30, ..., 110]`**
**How many "if-then" questions deep**

```
max_depth=3 → Shallow:
Carat>1? → Cut=Fair? → Color=E? → $2500 (3 questions)

max_depth=20 → Deep:
Carat>1.15? → Cut=Fair? → Color=E? → Carat>1.18? → ... → **$2510** (20 questions)
```
**Small=simple | Large=overfits (memorizes training data)**

### 3. **`learning_rate`: [0.1, 0.2, 0.3, ..., 0.9]`**
**How fast to learn from mistakes**

```
Tree1 predicts: $2000 (wrong by +500)

learning_rate=0.1  → Tree2 adds: 0.1×500 = **+50** → Total: $2050
learning_rate=0.5  → Tree2 adds: 0.5×500 = **+250** → Total: $2250
learning_rate=0.9  → Tree2 adds: 0.9×500 = **+450** → Total: $2450
```
**Small=slow learning (safer) | Large=fast learning (risky)**
